```python
import pandas as pd

gpt = pd.read_csv("/home/dbtproject/barsha/LLM/GPT_old.csv")
gemini = pd.read_csv("/home/dbtproject/barsha/LLM/Gemini_old.csv")

gpt = gpt.rename(
    columns={"predicted_class":"GPT_Pred"}
)

gemini = gemini.rename(
    columns={"predicted_class":"Gemini_Pred"}
)
```

```python

merged = pd.merge(
    gpt,
    gemini[["image_name","Gemini_Pred"]],
    on="image_name"
)

print(merged.head(20))
```

```python
merged["GPT_Correct"] = (
    merged["original_class"]
    ==
    merged["GPT_Pred"]
)

merged["Gemini_Correct"] = (
    merged["original_class"]
    ==
    merged["Gemini_Pred"]
)
```

```python
a = ((merged["GPT_Correct"] == True) &
     (merged["Gemini_Correct"] == True)).sum()

b = ((merged["GPT_Correct"] == True) &
     (merged["Gemini_Correct"] == False)).sum()

c = ((merged["GPT_Correct"] == False) &
     (merged["Gemini_Correct"] == True)).sum()

d = ((merged["GPT_Correct"] == False) &
     (merged["Gemini_Correct"] == False)).sum()
```

```python
print(f'both correct: {a}')
print(f'Only GPT correct: {b}')
print(f'only Gemini correct: {c}')
print(f'both incorrect: {d}')
```

```python
table = [
    [a, b],
    [c, d]
]
```

```python
import numpy as np
from statsmodels.stats.contingency_tables import mcnemar

result = mcnemar(
    table,
    exact=True
)

print(f'P-value : {result.pvalue: .6f}')
print("Statistic:", result.statistic)
```
