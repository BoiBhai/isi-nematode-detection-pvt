```python
PROMPT = """
You are an expert nematode taxonomist specializing in Meloidogyne species identification from microscopic perineal pattern images.

You will first observe reference images
belonging to three species:

1. graminicola
2. incognita
3. javanica

Study their morphology carefully.

Focus on:

- dorsal arch shape
- dorsal arch height
- lateral lines
- striae arrangement
- tail region
- overall perineal pattern

After observing the references,
analyze the final test image.

Predict which species it belongs to.

Return in this format:

Return the resultin this format :
match:Yes or No,
species:species name,
confidence:0-100,
reason:short explanation

"""
```

```python
import time
def predict_image(test_image_path):

    contents = [PROMPT]

    # Graminicola examples
    for img_path in support_examples["graminicola"]:

        contents.append(
            "Reference Example: graminicola"
        )

        contents.append(
            Image.open(img_path)
        )

    # Incognita examples
    for img_path in support_examples["incognita"]:

        contents.append(
            "Reference Example: incognita"
        )

        contents.append(
            Image.open(img_path)
        )

    # Javanica examples
    for img_path in support_examples["javanica"]:

        contents.append(
            "Reference Example: javanica"
        )

        contents.append(
            Image.open(img_path)
        )

    contents.append(
        "This is the test image."
    )

    contents.append(
        Image.open(test_image_path)
    )


    # Retry logic
    for attempt in range(5):

        try:

            response = client.models.generate_content(
                model="gemini-2.5-flash",
                contents=contents
            )

            return response.text

        except Exception as e:

            wait_time = 2 ** attempt

            print(
                f"Attempt {attempt+1} failed."
            )

            print(e)

            print(
                f"Waiting {wait_time} seconds..."
            )

            time.sleep(wait_time)

    return "API_ERROR"

    response = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=contents
    )

    return response.text
```

```python
def extract_species(output):

    try:

        data = json.loads(output)

        species = data["species"]

        return species.lower().strip()

    except:

        output = output.lower()

        if "graminicola" in output:
            return "graminicola"

        elif "incognita" in output:
            return "incognita"

        elif "javanica" in output:
            return "javanica"

        else:
            return "unknown"
```

```python
TEST_DIR = "/home/dbtproject/barsha/New_Image/test"

true_labels = []
pred_labels = []

results = []
```

```python
for class_name in os.listdir(TEST_DIR):

    class_path = os.path.join(
        TEST_DIR,
        class_name
    )

    for image_name in os.listdir(class_path):

        image_path = os.path.join(
            class_path,
            image_name
        )

        try:

            output = predict_image(
                image_path
            )

        except Exception as e:

            print("FAILED:", image_name)

            output = "API_ERROR"

        prediction = extract_species(
            output
        )

        if output == "API_ERROR":

            prediction = "unknown"

        true_labels.append(
            class_name
        )

        pred_labels.append(
            prediction
        )

        results.append({
            "image": image_name,
            "true": class_name,
            "predicted": prediction,
            "raw_output": output
        })

        pd.DataFrame(results).to_csv(
            "Fewshot_result.csv",
            index=False
        )

        print(
            image_name,
            prediction
        )

        time.sleep(3)
```
