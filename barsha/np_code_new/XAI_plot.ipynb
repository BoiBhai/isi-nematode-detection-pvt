```python
import cv2
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# Create Figure
# ============================================================
fig, axes = plt.subplots(
    nrows=6,
    ncols=4,
    figsize=(24, 34),
    constrained_layout=False
)

# ============================================================
# Display Images
# ============================================================
for row, img_paths in enumerate(image_sets):
    for col, img_path in enumerate(img_paths):

        img = Image.open(img_path).convert("RGB")
        img = np.array(img)

        axes[row, col].imshow(img)
        axes[row, col].axis("off")

# ============================================================
# Column Titles
# ============================================================
titles = [
    "Original Image",
    "Grad-CAM++",
    "Integrated Gradients (IG)",
    "LIME"
]

for col, title in enumerate(titles):
    axes[0, col].set_title(
        title,
        fontsize=18,
        fontweight="bold",
        pad=18
    )

# ============================================================
# Overall Figure Title
# ============================================================
fig.suptitle(
    "Explainability Results of ResNet50 on the Old Dataset",
    fontsize=24,
    fontweight="bold",
    y=0.985
)

# ============================================================
# Left-side Class Labels
# ============================================================
class_groups = [
    ("Class 0\nGraminicola", 0, 1),
    ("Class 1\nIncognita", 2, 3),
    ("Class 2\nJavanica", 4, 5),
]

nrows = 6

for label, start_row, end_row in class_groups:

    center = (start_row + end_row + 1) / 2
    y = 1 - center / nrows

    fig.text(
        0.070,             # move farther left
        y,
        label,
        rotation=90,
        fontsize=18,
        fontweight="bold",
        va="center",
        ha="center"
    )

# ============================================================
# Adjust Spacing
# ============================================================
plt.subplots_adjust(

    left=0.10,      # more space for class labels
    right=0.99,

    top=0.95,       # space for title
    bottom=0.03,

    wspace=0.06,    # horizontal spacing
    hspace=0.08     # vertical spacing
)

plt.show()
```

```python
from matplotlib.cm import ScalarMappable
from matplotlib.colors import Normalize

# Heatmap scale
norm = Normalize(vmin=0, vmax=1)

sm = ScalarMappable(
    cmap="jet",      # use the same cmap as GradCAM
    norm=norm
)
sm.set_array([])

# Shared colorbar
cbar = fig.colorbar(
    sm,
    ax=axes[:,1],        # attach to entire Grad-CAM++ column
    location="bottom",   # or "right"
    fraction=0.035,
    pad=0.03,
    aspect=40
)

cbar.set_label(
    "Importance Score",
    fontsize=14,
    fontweight="bold"
)

cbar.ax.tick_params(labelsize=11)
```
