In simple way

```python
x=torch.tensor(6.7)#input
y=torch.tensor(0.0)#true label

w=torch.tensor(1.0)#weight
b=torch.tensor(0.0)#bias
```

```python
#binary cross entropy loss for scaler
def binary_cross_entropy(Prediction, Target):
    epsilon = 1e-7  # small constant to avoid log(0)
    Prediction = torch.clamp(Prediction, epsilon, 1.0 - epsilon)  # clamp values to avoid log(0) 
    return - (Target * torch.log(Prediction) + (1 - Target) * torch.log(1 - Prediction))
```

```python
#forward pass
z= w*x + b  #linear function
y_pred = torch.sigmoid(z)  #activation function
loss = binary_cross_entropy(y_pred, y) #compute loss
```

```python
loss
```

```python
#backpropagation
dloss_dy_pred =(y_pred - y) / (y_pred * (1 - y_pred))
dy_pred_dz = y_pred * (1 - y_pred)

dz_dw = x
dz_db = 1

dl_dw = dloss_dy_pred * dy_pred_dz * dz_dw
dl_db = dloss_dy_pred * dy_pred_dz * dz_db
```

```python
print("Gradient w.r.t weight:", dl_dw)
print("Gradient w.r.t bias:", dl_db)
```
